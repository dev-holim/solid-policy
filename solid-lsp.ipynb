# 객체지향 5원칙 (SOLID)

## LSP 리스코프 치환 원칙


##### 하위 클래스는 상위 클래스의 기능을 변경하지 않고 확장해야 한다.
##### 상위 타입의 객체를 하위 타입 객체로 대체해도 정상적으로 동작해야 한다.

---

```python
# Penguin은 Bird를 상속받았지만, `fly()` 메서드를 사용할 수 없음 (LSP 위반)

class Bird:
    def fly(self):
        return "Flying"

class Penguin(Bird):
    def fly(self):
        raise Exception("Penguins cannot fly")
```

```python
# Penguin은 적절한 부모 클래스를 상속받아 move() 메서드를 올바르게 동작하도록 구현 (LSP 준수)

from abc import ABC, abstractmethod

class Bird(ABC):
    @abstractmethod
    def move(self):
        pass

class FlyingBird(Bird):
    def move(self):
        return "Flying"

class NonFlyingBird(Bird):
    def move(self):
        return "Walking"

class Penguin(NonFlyingBird):
    pass
```
