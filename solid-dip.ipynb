## DIP 의존 역전 원칙

##### 고수준 모듈이 저수준 모듈에 의존하지 않고, 추상화에 의존해야 한다.
##### 의존성을 직접 사용하지 않고 인터페이스를 통해 주입해야 한다.


![nn](./img/solid-dip.png)

---

```python
# DIP 위반

class Keyboard:
    def input(self):
        return "User input"

class Computer:
    def __init__(self):
        self.keyboard = Keyboard()  # 직접 객체 생성 (DIP 위반)

    def read_input(self):
        return self.keyboard.input()
```

```python
# 의존성 주입을 통해 Computer는 Keyboard가 아닌 추상 InputDevice에 의존 (DIP 준수)

from abc import ABC, abstractmethod

class InputDevice(ABC):
    @abstractmethod
    def input(self):
        pass

class Keyboard(InputDevice):
    def input(self):
        return "User input"

class Computer:
    def __init__(self, input_device: InputDevice):
        self.input_device = input_device  # 의존성 주입 (DIP 준수)

    def read_input(self):
        return self.input_device.input()


keyboard = Keyboard()
computer = Computer(keyboard)
```
