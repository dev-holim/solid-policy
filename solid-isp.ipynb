## ISP 인터페이스 분리 원칙

##### 클라이언트는 자신이 사용하지 않는 기능에 의존하지 않아야 한다.
##### 인터페이스는 세분화되어야 하며, 사용하지 않는 메서드를 강요하지 않아야 한다.

---

```python
# Robot이 eat() 메서드를 구현해야 하는 문제 발생 (ISP 위반)

class Worker:
    def work(self):
        pass

    def eat(self):
        pass

class Engineer(Worker):
    def work(self):
        return "Coding"

    def eat(self):
        return "Eating"

class Robot(Worker):
    def work(self):
        return "Automating"

    def eat(self):
        raise Exception("Robots do not eat!")
```

```python
# Robot이 eat() 메서드를 강제하지 않음 (ISP 준수)

from abc import ABC, abstractmethod

class Workable(ABC):
    @abstractmethod
    def work(self):
        pass

class Eatable(ABC):
    @abstractmethod
    def eat(self):
        pass

class Engineer(Workable, Eatable):
    def work(self):
        return "Coding"

    def eat(self):
        return "Eating"

class Robot(Workable):
    def work(self):
        return "Automating"
```
