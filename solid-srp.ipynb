## SRP 단일 책임 원칙


##### 한 클래스는 하나의 책임만 가져야 한다.
##### 클래스는 단 하나의 기능을 수행해야 하며, 변경의 이유가 오직 하나여야 한다.

---

```python
# 기존 원칙 위반 코드
'''
Report 클래스가 `보고서 생성`과 `파일 저장`이라는 두 가지 책임을 가짐 (SRP 위반)
'''

class Report:
    def __init__(self, data):
        self.data = data

    def generate_report(self):
        return f"Report Data: {self.data}"

    def save_to_file(self, filename):
        with open(filename, 'w') as file:
            file.write(self.generate_report())
```

```python
# 수정 코드
'''
Report 클래스는 "보고서 생성"만 담당하고, ReportSaver는 "파일 저장"만 담당 (SRP 준수)
'''

class Report:
    def __init__(self, data):
        self.data = data

    def generate_report(self):
        return f"Report Data: {self.data}"

class ReportSaver:
    @staticmethod
    def save_to_file(report: Report, filename):
        with open(filename, 'w') as file:
            file.write(report.generate_report())
```
